# Random Tensors
### Why Random Tensor?
#### Neural Networks starts with a tensor with random numbers and learns by updating the random numbers again and again

```python
random_tensor = torch.rand(1,4,5)
random_tensor, random_tensor.ndim , random_tensor.shape
```
