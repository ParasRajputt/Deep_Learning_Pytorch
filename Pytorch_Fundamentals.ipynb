Chapter 0 — Tensor Fundamentals: Notes



1.  ndim: nesting depth of a tensor (how many bracket levels before a raw number)

2.  shape: element count at each nesting level, outer→inner; len(shape) == ndim

3. Zero-init problem: all neurons get identical gradients → identical updates forever → network collapses to one effective neuron. Random init breaks symmetry.

4. float32 vs float16: float32 = default, more precision, more memory, slower. float16 = half memory, faster on Tensor Core GPUs (T4), but risks underflow on small values (e.g. small gradients rounding to 0). Mixed precision training uses float16 compute + float32 master weights to get both benefits.

5. matmul (2D): A(n×m) @ B(x×y) valid only if m==x; result shape (n×y). Each element = row·column dot product. 1D tensors matmul as a dot product (single scalar), not row/col rules.

6. reshape: total element count must match old and new shape. .view() requires contiguous memory and shares memory with original (no copy) — mutating one mutates the other. .reshape() uses view if possible, else copies.
squeeze/unsqueeze: remove/add size-1 dimensions to match shape expectations between tensors, ops, or models (e.g. adding a batch dim before feeding a single image to a model).

7. Indexing: x[row] = row, x[row,col] = element, x[:,col] = column.

8. NumPy interop: torch.from_numpy() / .numpy() share memory — same mutation risk as .view().

9. torch.manual_seed(n): fixes the RNG's starting point for the next call only, not permanently. Re-seed before each call if you need identical outputs from separate calls.
