```python
import torch
from torch import nn

import torchvision
from torchvision import datasets
from torchvision.transforms import ToTensor
```

```python
train_data = datasets.FashionMNIST(
    root = 'data',
    train = True ,
    transform = ToTensor(),
    download = True ,
    target_transform = None
)
test_data = datasets.FashionMNIST(
    root = 'data',
    train = False ,
    download = True,
    transform = ToTensor()
)
```

```python
from torch.utils.data import DataLoader
train_dataloader = DataLoader(train_data,
                              batch_size = 32,
                              shuffle = True)
test_dataloader = DataLoader(test_data,
                             batch_size = 32,
                             shuffle = False)

print(f"Length of train dataloader: {len(train_dataloader)} of batch 32")
print(f"Lenght of test dataloader : {len(test_dataloader)} of batch 32")
```

```python
class CNN_MODEL(nn.Module):
  def __init__(self , IN_CHANNELS:int , OUT_CHANNELS:int , HIDDEN_CHANNELS:int):
    super().__init__()
    self.block_1 = nn.Sequential(
        nn.Conv2d(in_channels = IN_CHANNELS,
                  out_channels = HIDDEN_CHANNELS,
                  kernel_size = 3,
                  stride = 1,
                  padding = 1),
        nn.ReLU(),
        nn.Conv2d(in_channels = HIDDEN_CHANNELS,
                  out_channels = HIDDEN_CHANNELS,
                  kernel_size = 3,
                  stride = 1,
                  padding = 1),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size = 2,
                     stride = 2,
                    )

    )
    self.block_2 = nn.Sequential(
        nn.Conv2d(HIDDEN_CHANNELS , HIDDEN_CHANNELS , 3 , 1 , 1),
        nn.ReLU(),
        nn.Conv2d(HIDDEN_CHANNELS , HIDDEN_CHANNELS , 3,1,1),
        nn.ReLU(),
        nn.MaxPool2d(2 , 2 )
    )
    self.classifier= nn.Sequential(
        nn.Flatten(),
        nn.Linear(in_features = HIDDEN_CHANNELS*7*7,
                  out_features = OUT_CHANNELS)
    )

  def forward(self , x):
    x = self.block_1(x)
    x = self.block_2(x)
    x = self.classifier(x)

    return x
```

```python
torch.manual_seed(42)
model = CNN_MODEL(IN_CHANNELS = 1,
                  HIDDEN_CHANNELS = 10,
                  OUT_CHANNELS = len(train_data.classes))
model
```

```python
from torch.optim import Adam
```

```python
loss_fn = nn.CrossEntropyLoss()
optimizer = Adam(model.parameters(), lr=0.001)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
```

```python
def train_step(model, dataloader, loss_fn, optimizer, device):
    model.train()
    train_loss, train_acc = 0, 0

    for X, y in dataloader:
        X, y = X.to(device), y.to(device)

        # Forward pass
        y_pred = model(X)

        # Loss and accuracy
        loss = loss_fn(y_pred, y)
        train_loss += loss.item()
        y_pred_class = torch.argmax(torch.softmax(y_pred, dim=1), dim=1)
        train_acc += (y_pred_class == y).sum().item() / len(y)

        # Optimizer zero grad
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Optimizer step
        optimizer.step()

    train_loss /= len(dataloader)
    train_acc /= len(dataloader)
    return train_loss, train_acc

def test_step(model, dataloader, loss_fn, device):
    model.eval()
    test_loss, test_acc = 0, 0

    with torch.inference_mode():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)

            # Forward pass
            test_pred = model(X)

            # Loss and accuracy
            loss = loss_fn(test_pred, y)
            test_loss += loss.item()
            test_pred_class = torch.argmax(torch.softmax(test_pred, dim=1), dim=1)
            test_acc += (test_pred_class == y).sum().item() / len(y)

    test_loss /= len(dataloader)
    test_acc /= len(dataloader)
    return test_loss, test_acc
```

```python
torch.manual_seed(42)
epochs = 3

for epoch in range(epochs):
    print(f"Epoch: {epoch+1}\n---------")
    train_loss, train_acc = train_step(model, train_dataloader, loss_fn, optimizer, device)
    test_loss, test_acc = test_step(model, test_dataloader, loss_fn, device)

    print(f"Train loss: {train_loss:.4f} | Train acc: {train_acc*100:.2f}%")
    print(f"Test loss: {test_loss:.4f} | Test acc: {test_acc*100:.2f}%\n")
```

```python
MODEL_PATH = "cnn_fashion_model.pth"
torch.save(model.state_dict(), MODEL_PATH)
print(f"Model successfully saved to {MODEL_PATH}")
```

```python
loaded_model = CNN_MODEL(
    IN_CHANNELS=1,
    HIDDEN_CHANNELS=10,
    OUT_CHANNELS=len(train_data.classes)  #[cite: 1]
)
loaded_model.load_state_dict(torch.load(MODEL_PATH))

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
loaded_model.to(device)
loaded_model.eval()
print("Model reloaded into a fresh instance and moved to device.")
```

```python
single_image, single_label = test_data[100]  # Grab the first test image/label[cite: 1]
single_image_tensor = single_image.unsqueeze(0).to(device)  # Add batch dimension [1, 1, 28, 28]

with torch.inference_mode():
    single_pred_logits = loaded_model(single_image_tensor)
    single_pred_idx = torch.argmax(single_pred_logits, dim=1).item()

predicted_class_name = train_data.classes[single_pred_idx]  #[cite: 1]
actual_class_name = train_data.classes[single_label]          #[cite: 1]

print(f"\n--- Single Image Prediction Test ---")
print(f"Predicted Class: {predicted_class_name}")
print(f"Actual Class:    {actual_class_name}")
```
